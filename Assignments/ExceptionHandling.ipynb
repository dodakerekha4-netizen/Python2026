```python
# 1. Write a Python program to handle a ZeroDivisionError exception when dividing a number by zero.

def divide(numerator, denominator):
    try:
        # Code that might raise an exception
        result = numerator / denominator
        print(f"Result: {result}")
    except ZeroDivisionError:
        # Code to execute if a ZeroDivisionError occurs
        print("Error: Cannot divide by zero.")


# Usage examples
print("division is 15 / 2:")
divide(15, 2)

print("\n division of 10 / 0:")
divide(10, 0)

```

```python
# 3. Write a Python program that opens a file and handles a FileNotFoundError exception if the file does not exist.

def file_write(sample):
  try:
   with open('sample.txt','r') as f:
    f.read
  except FileNotFoundError:
   print("file is not found")

# call function
  file_write(sample.txt)
```

```python
# 5. Write a Python program that opens a file and handles a PermissionError exception if there is a permission issue.
def open_file(filename):
  try:
    with open(filename,"r") as f:
      content = f.read()
      print("file content :\n :",content)
  except PermissionError:
    print(f"error: open file permission issue '{filename}' :")

    # call function
    filename = input("Enter the filename to open: ")
    open_file(filename)


```

```python
# 6. Write a Python program that executes an operation on a list and handles an IndexError exception if the index is out of range.

def index(data_list,index):

  try:
    result = data_list[index]
    print(f"The element at index {index} is: {result}")
  except IndexError:
        # Handle the exception if the index is out of bounds
        print(f"Error: The index {index} is out of range for this list.")



my_list = [10, 20, 30, 40, 50]

# Test with a valid index
index(my_list, 2)

# Test with an out-of-range index
index(my_list, 10)
```

```python
# 7. Write a Python program that executes division and handles an ArithmeticError exception if there is an arithmetic error.
def safe_division(dividend, divisor):
    try:
        # Attempt to divide the numbers
        result = dividend / divisor
        print(f"Result: {result}")
    except ArithmeticError as e:
        # Catches ZeroDivisionError, OverflowError, etc.
        print(f"Error: An arithmetic error occurred. ({e})")

# Usage Examples
print("Test 1: Normal Division")
safe_division(10, 2)

print("\nTest 2: Division by Zero")
safe_division(10, 0)
```
