```python
# 5. You are given a list of integers. You are asked to make a list by running
  # through elements of the list by adding all elements greater and itself.
  # i.e. Say given list is [2,4,6,10,1] resultant list will be [22,20,10,23].
# For 1st element 2 ->> these are greater (4+6+10) values and 2 itself so on adding becomes 22.
# For 2nd element 4 ->> greater elements are (6, 10) and 4 itself, so on adding 20
# like wise for all other elememts.
# [2,4,6,10,1]-->[22,20,16,10,23]


#list1 = [2,4,6,10,1]

def greater_sum_list(nums):
    result = []
    for i in nums:
        total = i + sum(x for x in nums if x > i)
        result.append(total)
    return result

# Demo
nums = [2, 4, 6, 10, 1]
print(greater_sum_list(nums))

```

```python
# 6. Find list of common unique items from two list. and show in increasing order
# Input
# num1 = [23,45,67,78,89,34]
# num2 = [34,89,55,56,39,67]
# Output:
# [34, 67, 89]

def common_unique_sorted(num1, num2):
    result = []
    for i in num1:
      if i in num2 and i not in result:  # common + unique
          result.append(i)
    result.sort()  # increasing order
    return result

# Demo
num1 = [23, 45, 67, 78, 89, 34]
num2 = [34, 89, 55, 56, 39, 67]
print(common_unique_sorted(num1, num2))  # Output: [34, 67, 89]

# without function

num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
result = []
for i in num1:
  if i in num2 and i not in result:
    result.append(i)
    result.sort()
    print(result)
```

```python
# 7. Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.
#	Input:
#	['1ac21', '23fg', '456', '098d','1','kls']
#	Output:
#	['456', '23fg', '1ac21', '1', 'kls', '098d']

def product_of_digits(s):
    digits = [int(ch) for ch in s if ch.isdigit()]
    if not digits:   # no digits → product = 1
        return 1
    product = 1
    for d in digits:
        product *= d
    return product

def sort_by_product(strings):
    return sorted(strings, key=product_of_digits, reverse=True)

# Input
data = ['1ac21', '23fg', '456', '098d', '1', 'kls']

# Output
result = sort_by_product(data)
print(result)
```

```python
# 13. Convert a list of Tuples into Dictionary.
# Example 1:
# Input:
#	[("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
#	Output:
# {'akash': [10], 'gaurav': [12], 'anand': [14], 'suraj': [20], 'akhil': [25], 'ashish': [30]}
# Example 2:
# Input:
# [('A', 1), ('B', 2), ('C', 3)]
#	Output:
#	{'A': [1], 'B': [2], 'C': [3]}


def convert_to_dict(tuple_list):
    return {key: [value] for key, value in tuple_list}

# Example 1
input1 = [("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
print(convert_to_dict(input1))

# Example 2
input2 = [('A', 1), ('B', 2), ('C', 3)]
print(convert_to_dict(input2))
```

```python
# 16. Write a Python function to check whether a number is perfect or not.¶
#	A Perfect number is a number that is half the sum of all of its positive divisors (including itself).
#	Example :
#	The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
#	Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.
#	The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.


def is_perfect(n):
    """
    Checks if a number n is a perfect number.
    """
    if n < 2:
        return False

    # Initialize sum of divisors
    sum_divisors = 0

    # Iterate through potential divisors up to n/2
    for i in range(1, n // 2 + 1):
        if n % i == 0:
            sum_divisors += i

    # Return True if sum of divisors equals the number
    return sum_divisors == n

# Example
print(is_perfect(6))
print(is_perfect(28))
print(is_perfect(12))
```

```python
# 17. Write a Python function to concatenate any no of dictionaries to create a new one.
# Sample Dictionary :
#	dic1={1:10, 2:20}
#	dic2={3:30, 4:40}
#	dic3={5:50,6:60}
#	Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}


def concatenate_dicts(*dicts):
    result = {}
    for d in dicts:
        result.update(d)
    return result

# Sample Dictionaries
dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}

# Call function
new_dict = concatenate_dicts(dic1, dic2, dic3)
print(new_dict)
```

```python
# 18. Write a python function that receives a list of integers and prints out a histogram of bin size 10
# Input:
#	[13,42,15,37,22,39,41,50]
#	Output:

#	{11-20:2,21-30:1,31-40:2,41-50:3}

def print_histogram(data):
    if not data:
        return {}

    # Sort the data to easily find the range
    sorted_data = sorted(data)

    # Calculate the start of the first bin (e.g., 11 for 13) and end
    start = (sorted_data[0] // 10) * 10
    if sorted_data[-1] % 10 == 0:
        end = sorted_data[-1]
    else:
        end = ((sorted_data[-1] // 10) + 1) * 10

    histogram = {}

    # Iterate through bins in steps of 10
    for i in range(start, int(end), 10):
        bin_label = f"{i+1}-{i+10}"
        # Count elements that fall into the current bin range
        count = len([x for x in sorted_data if i < x <= i + 10])
        histogram[bin_label] = count

    print(histogram)

# Input data
input_list = [13, 42, 15, 37, 22, 39, 41, 50]
print_histogram(input_list)
```
