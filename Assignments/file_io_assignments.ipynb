```python
# 3. Write a Python program to append text to a file and display the text.

filename = "example.txt"
text_to_append = "Basically I am from Pune"

with open(filename,"r") as file:
 content = file.read()

 if text_to_append not in content:
    with open(filename, "a") as file:
        file.write(text_to_append + "\n")

with open (filename,"r") as f:
 lines = f.read()
 print(lines)
```

```python
# 4. Write a Python program to read last n lines of a file.

def last_line(filename, n):
    with open(filename,'r') as file:
        content = file.readlines()
        last_line = content[-n:]
        return last_line


filename = "pythondef.txt"
n = 1
print(f"Last {n} lines of {filename}:\n")
print(last_line(filename, n))

```

```python
# 7. Write a python program to find the longest words.

s = "I am learning Python programming language"

# Split the sentence into words
words = s.split()

# Find the longest word
res = max(words, key=len)

print(res)

```

```python
# 9. Write a Python program to count the frequency of words in a file.

count = 0

# Opens a file in read mode
f = open("example.txt", "r")

# Gets each line till end
for line in f:
    # Splits into words
    word = line.split(" ")
    # Counts each words
    count += len(word)

print("Total Number of Words: " + str(count))
f.close()
```

```python
# 10. Write a Python program to get the file size of a plain file.

count = 0
file = open('example.txt','r')

for line in file:
  word = line.split(" ")
  count += len(word)
print("Size of file is :", file.tell(), "bytes")

```

```python
#15.Write a Python program that takes a text file as input and returns the number of words of a given text file.

word_count = 0

filename = "pythondef.txt"

with open(filename, 'r') as file:

    for fileline in file:
      # split the line into words using split() function.
      # increasing the word count by counting the number of words in the file
        word_count += len(fileline.split())
print("The total number of files present in the given file = ", word_count)
```
